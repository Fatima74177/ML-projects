# E-Commerce Product Data Analysis

This notebook runs the scraper, cleans and validates the dataset, exports CSV and Excel files, and creates the requested visualizations.

```python
from scraper.scraper import Scraper
from scraper.processing import clean_products, validate_products
from scraper.analysis import summarize, create_visualizations
from scraper.storage import save_outputs
from scraper.config import OUTPUT_DIR

records = Scraper().scrape()
products = clean_products(records)
valid, errors = validate_products(products)
print(valid, errors, products.shape)
```

```python
summary = summarize(products)
summary
```

```python
csv_path, excel_path = save_outputs(products, OUTPUT_DIR)
chart_paths = create_visualizations(products, OUTPUT_DIR)
print(csv_path, excel_path, chart_paths)
```
